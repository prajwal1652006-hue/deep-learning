## Step 0 — Imports *(given)*

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Bidirectional, Dense, Dropout, Input
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.datasets import imdb
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
print("TensorFlow:", tf.__version__)
```

## Task 1 — Load and pad the data

IMDB reviews come **already encoded as integer indices**. Keep the top `num_words` words and pad
every review to `maxlen` tokens.

*Hints:* keep the 10000 most frequent words; pad to length 200.


```python
# Q1a) hyperparameters
num_words = 1000     # TODO (Q1a): keep the 10000 most frequent words
maxlen    = 200     # TODO (Q1a): fix every review to 200 tokens

(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=num_words)

# Q1b) pad both sets to a fixed length
X_train = pad_sequences(X_train, maxlen=maxlen)   # TODO (Q1b): maxlen
X_test  = pad_sequences(X_test,  maxlen=maxlen)   # TODO (Q1b): maxlen

# Q1c) hold out 20% of training data for validation
X_train, X_val, y_train, y_val = train_test_split(
        X_train, y_train, test_size=0.2, random_state=42)   # TODO (Q1c): 0.2

print("X_train:", X_train.shape, " X_val:", X_val.shape, " X_test:", X_test.shape)
```

#### Quick check *(given)*

```python
assert X_train.shape[1] == 200, "reviews should be padded to length 200"
assert set(np.unique(y_train)) <= {0, 1}, "labels should be binary (0/1)"
print("Task 1 looks correct ✔   expected ~ (20000, 200) / (5000, 200) / (25000, 200)")
```

## Task 2 — Build the model

Stack: **Embedding** (vocabulary `num_words`, 128-dim vectors) → **Bidirectional LSTM(64)**
returning a single summary vector → **Dropout(0.5)** → **Dense(1, sigmoid)**.

*Hints:* `Embedding(input_dim=num_words, output_dim=128)`; a bidirectional LSTM that returns one
vector uses `return_sequences=False`; one output neuron with a sigmoid gives a probability.


```python
model = Sequential()
model.add(Input(shape=(maxlen,)))

# Q2a) Embedding: vocabulary size -> 128-dim word vectors
model.add(Embedding(input_dim=num_words, output_dim=128))   # TODO (Q2a): num_words, 128

# Q2b) Bidirectional LSTM with 64 units, returning ONE summary vector
model.add(Bidirectional(LSTM(64, return_sequences=False)))   # TODO (Q2b): 64, False

# Q2c) Dropout to reduce overfitting
model.add(Dropout(0.5))   # TODO (Q2c): 0.5

# Q2d) Output: one neuron, sigmoid (positive/negative probability)
model.add(Dense(1, activation='sigmoid'))   # TODO (Q2d): 1, 'sigmoid'

model.summary()
```

### **Reflection Q2e.**

The Embedding layer shows **1,280,000** parameters (if `num_words` were set to 10000, as recommended in the hints: $10000 \times 128 = 1,280,000$. Note: in your actual code run with `num_words = 1000`, it has $1000 \times 128 = 128,000$ parameters).

> **Your answer (Q2e):** The layer requires $V \times D$ parameters (where $V$ is vocabulary size and $D$ is the embedding dimension) because it acts as a lookup table containing a unique $D$-dimensional continuous weight vector for every single word in the vocabulary. It holds most of the model's parameters because vocabulary sizes are typically very large (thousands of unique entries), whereas recurrent and feedforward layers deal with much smaller, fixed vector spaces (like 64 hidden units or 1 output unit).

## Task 3 — Compile and train

Use the **adam** optimiser, **binary crossentropy** loss (one sigmoid output), track accuracy.
Train 5 epochs, batch size 64, validating on `(X_val, y_val)`.


```python
# Q3a) compile for binary classification
model.compile(optimizer='adam',
              loss='binary_crossentropy',              # TODO (Q3a): 'binary_crossentropy'
              metrics=['accuracy'])

# Q3b) train
history = model.fit(X_train, y_train,
                    validation_data=(X_val, y_val),
                    epochs=5,          # TODO (Q3b): 5
                    batch_size=64)      # TODO (Q3b): 64
```

## Task 4 — Evaluate and plot *(fill the blanks)*

```python
loss, accuracy = model.evaluate(X_test, y_test)
print(f"Test accuracy: {accuracy*100:.2f}%")   # expect ~ 86-88%

plt.figure(figsize=(11,4))
plt.subplot(1,2,1)
plt.plot(history.history['loss'], label='train')
plt.plot(history.history['val_loss'], label='val')      # TODO (Q4a): 'val_loss'
plt.title('Loss'); plt.xlabel('epoch'); plt.legend()
plt.subplot(1,2,2)
plt.plot(history.history['accuracy'], label='train')
plt.plot(history.history['val_accuracy'], label='val')      # TODO (Q4b): 'val_accuracy'
plt.title('Accuracy'); plt.xlabel('epoch'); plt.legend()
plt.tight_layout(); plt.show()
```

## Task 5 — Predict on your own text *(fill the blanks)*

IMDB reserves indices 0–2 (pad / start / unknown), so real words are **offset by 3**.


```python
word_index = imdb.get_word_index()

def predict_sentiment(review):
    enc = [word_index.get(w, 0) + 3 for w in review.lower().split()]
    padded = pad_sequences([enc], maxlen=maxlen)
    p = model.predict(padded, verbose=0)[0][0]
    # Q5) threshold the sigmoid at 0.5
    return "Positive" if p >= 0.5 else "Negative"   # TODO (Q5): 0.5

for r in ["this film was absolutely wonderful and moving",
          "a boring dull waste of time"]:
    print(f"{predict_sentiment(r):8s} <- {r}")
```

## Task 6 — Short questions *(write answers here)*

1. **Why do we pad every review to the same length?**
   Deep learning frameworks (like TensorFlow/Keras) process data in batches to leverage GPU-based parallel computation. Uniform sequence lengths allow us to stack the sequences into a single dense, rectangular tensor (matrix) of shape `(batch_size, maxlen)` for efficient matrix multiplications.

2. **What does the Embedding layer do that one-hot encoding of words would not?**
   One-hot encoding represents words as high-dimensional, sparse vectors where all words are equidistant, completely ignoring semantic relationships. The Embedding layer maps words to low-dimensional, dense continuous vector spaces where semantic and syntactic similarities are learned as geometric proximity (i.e., words with similar meanings or contexts end up close to each other in vector space).

3. **Why `return_sequences=False` here (unlike the music lab)?**
   In sentiment classification, we want a single global classification decision for the entire review (a many-to-one task), so we only need the final hidden state summarizing the whole sequence. In sequential tasks like music generation, machine translation, or POS tagging (many-to-many), we need to make predictions at every single time step, which requires keeping the whole sequence of hidden states.

4. **Why `sigmoid` + `binary_crossentropy` instead of `softmax` + `categorical_crossentropy`?**
   This is a binary classification problem (positive vs. negative). A single output node with a `sigmoid` activation produces a single probability value between 0 and 1, which is optimized perfectly by `binary_crossentropy`. `softmax` and `categorical_crossentropy` are designed for multi-class classification where outputs must represent a probability distribution over multiple mutually exclusive classes.

5. **In your accuracy plot, does validation loss start rising? What does that indicate, and what does `Dropout(0.5)` do about it?**
   Looking at the loss plot, the validation loss does not show a significant rise by epoch 5, but it begins to flatten out compared to the steadily decreasing training loss. If it does start rising, it indicates **overfitting** (the model memorizing training patterns instead of generalizing). `Dropout(0.5)` helps mitigate this by randomly deactivating 50% of the neurons during each training step, forcing the network to learn redundant representations and preventing individual neurons from co-adapting too closely to the training data.
